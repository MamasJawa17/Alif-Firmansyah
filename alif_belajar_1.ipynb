```python
panjang = float(input("panjang (m): "))
lebar = float(input("lebar (m): "))
tinggi = float(input("tinggi (m): "))

volume = panjang * lebar * tinggi

print("volume =", volume, "m3")
```

```python
panjang = 5.0
lebar = 0.25
tinggi = 0.40

volume = panjang * lebar * tinggi

print("volume balok =", volume, "m3")
```

```python
volume = float(input("volume beton (m3): "))
if volume >= 10:
  print("volume besar")
elif volume >= 5:
  print("volume sedang")
else:
  print("volume kecil")
```

```python
volume_satu_balok = 0.5
jumlah_balok = 4
total_volume = 0

for i in range(jumlah_balok):
  total_volume += volume_satu_balok

print("total volume =", total_volume, "m3")
```

```python
volume = [35, 42, 48, 45]
total = sum(volume)
rata_rata = sum(volume) / len(volume)
tertinggi = max(volume)
terendah = min(volume)

print("total:", total)
print("rata-rata:", rata_rata)
print("tertinggi:", tertinggi)
print("terendah:", terendah)
```
