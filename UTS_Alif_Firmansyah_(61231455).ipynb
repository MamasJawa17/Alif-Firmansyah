```python
import pandas as pd

data = {
    "waktu": ["07.00", "08.00", "09.00", "10.00", "11.00", "12.00", "13.00", "14.00", "15.00", "16.00"],
        "Volume": [120, 185, 150, 110, 135, 160, 145, 125, 175, 190],
    }

df = pd.DataFrame(data)
print(df)
```

```python
import numpy as pd

volume = pd.array([120, 185, 150, 110, 135, 160, 145, 125, 175, 190])
print("total Volume Kendaraan:", pd.sum(volume))
print("rata-rata Volume Kendaraan:", pd.mean(volume))
print("Volume Kendaraan Tertinggi:", pd.max(volume))
print("Volume Kendaraan Terendah:", pd.min(volume))
```

keismpulan volume diatas rata-rata terjadi di jam 08.00, jam 09.00, Jam 12.00, Jam 15.00, Jam 16.00, menyebabkan kemungkinan kemacetan
